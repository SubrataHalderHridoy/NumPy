```python
import numpy as np
```

```python
### Common Data Types in Numpy
#        - Integers: int32, int64
#        - Floating point nums: float32, float64
#        - Boolean: bool
#        - Complex nums: complex64, complex128
#        - String: S (byte-str) & U (unicode-str)
#        - Object: generic python objects - object
```

```python
# Integers

arr1 = np.array([1, 2, 3, 4, 5])

print(arr1, arr1.dtype)
```

```python
# Float

arr2 = np.array([1, 2, 3, 4.5, 5])

print(arr2, arr2.dtype)
```

```python
# bool

arr4 = np.array([True, False, True], dtype=bool)

print(arr4 ,arr4.dtype)
```

```python
# Complex nums

arr1 = np.array([2 + 3j])
arr2 = np.array([5 + 8j])

print(arr1, arr1.dtype)
```

```python
# Data types

arr = np.array(["hridoy", {1, 2, 3}, 3.14])

print(arr, arr.dtype)
```
